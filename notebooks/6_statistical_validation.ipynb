#### Goal
Validate the visual trends observed in previous notebooks using rigorous statistical testing.

**Tests Performed:**
1. **Two-Sample T-Test:** Remote (100%) vs. On-Site (0%) Salaries.
2. **Two-Sample T-Test:** US-based vs. International Salaries.
3. **One-Way ANOVA:** Salary differences across Company Sizes (S, M, L).

**Significance Level:** $\alpha = 0.05$ (5%)

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import seaborn as sns
import matplotlib.pyplot as plt

# Configuration
pd.set_option('display.float_format', '{:.4f}'.format)

# Load Data
df = pd.read_csv('../data/ds_salaries_cleaned.csv')

print("Data Loaded for Statistical Testing.")
```

```python
#  Isolate the groups
remote_salaries = df[df['remote_ratio'] == 100]['salary_in_usd']
onsite_salaries = df[df['remote_ratio'] == 0]['salary_in_usd']

#  Perform T-Test (assuming unequal variance is safer)
t_stat, p_value = stats.ttest_ind(remote_salaries, onsite_salaries, equal_var=False)

#  Interpret Results
print(f"Remote Mean Salary: ${remote_salaries.mean():,.2f}")
print(f"On-Site Mean Salary: ${onsite_salaries.mean():,.2f}")
print(f"\nT-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}")

if p_value < 0.05:
    print("\nResult: REJECT Null Hypothesis. The salary difference is Statistically Significant.")
else:
    print("\nResult: FAIL TO REJECT Null Hypothesis. No significant difference found.")
```

```python
#  Create a binary column for US vs Non-US
df['is_US'] = df['company_location'].apply(lambda x: 'US' if x == 'US' else 'Rest of World')

#  Isolate groups
us_salaries = df[df['is_US'] == 'US']['salary_in_usd']
global_salaries = df[df['is_US'] == 'Rest of World']['salary_in_usd']

#  Perform T-Test
t_stat_us, p_val_us = stats.ttest_ind(us_salaries, global_salaries, equal_var=False)

print(f"US Mean Salary: ${us_salaries.mean():,.2f}")
print(f"Global Mean Salary: ${global_salaries.mean():,.2f}")
print(f"Difference: ${us_salaries.mean() - global_salaries.mean():,.2f}")
print(f"\nP-Value: {p_val_us:.4e}") # Scientific notation for very small numbers

if p_val_us < 0.05:
    print("\nConclusion: The 'US Premium' is real and statistically significant.")
```

```python
#  Isolate the 3 groups
salary_S = df[df['company_size'] == 'S']['salary_in_usd']
salary_M = df[df['company_size'] == 'M']['salary_in_usd']
salary_L = df[df['company_size'] == 'L']['salary_in_usd']

# Perform ANOVA
f_stat, p_value = stats.f_oneway(salary_S, salary_M, salary_L)

print(f"Small Company Mean: ${salary_S.mean():,.0f}")
print(f"Medium Company Mean: ${salary_M.mean():,.0f}")
print(f"Large Company Mean:  ${salary_L.mean():,.0f}")

print(f"\nANOVA F-Statistic: {f_stat:.2f}")
print(f"P-Value: {p_value:.4f}")

if p_value < 0.05:
    print("\nResult: Significant difference found between company sizes.")
else:
    print("\nResult: No significant difference detected.")
```
