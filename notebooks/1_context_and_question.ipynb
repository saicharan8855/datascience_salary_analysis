## 4. Strategic Questions & Hypotheses
This analysis is guided by the following core questions:

**1. The "Seniority Premium" Analysis**
> *Hypothesis:* The salary gap between Senior (SE) and Executive (EX) roles is statistically significant, but the gap between Entry (EN) and Mid (MI) is narrowing due to demand for junior talent.

**2. The Remote Work Arbitrage**
> *Hypothesis:* Fully remote roles (100) command a higher average salary than hybrid roles (50) due to the global competition for talent, effectively decoupling salary from local cost-of-living.

**3. Geographic Dominance**
> *Hypothesis:* US-based companies pay significantly higher salaries (>$50k premium) compared to the global average, even for the same job titles.

**4. Temporal Inflation**
> *Hypothesis:* Salaries for "Data Engineers" have grown faster year-over-year (2020-2022) than "Data Scientists" due to the shifting focus towards infrastructure and MLOps.
