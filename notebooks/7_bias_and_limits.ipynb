#### Goal
Critically assess the reliability of our findings by identifying biases in the dataset.

**Identified Biases:**
1.  **Geographic Bias:** The dataset is heavily skewed towards US and UK markets.
2.  **Experience Bias:** Entry-level roles are underrepresented compared to Senior roles.
3.  **Currency Volatility:** Salaries converted to USD do not account for Purchasing Power Parity (PPP).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

df = pd.read_csv('../data/ds_salaries_cleaned.csv')
print("Data Loaded for Bias Analysis.")
```

```python
# Calculate country representation
country_counts = df['company_location'].value_counts().head(10)

plt.figure(figsize=(12, 6))

# barplot
sns.barplot(x=country_counts.values, y=country_counts.index, 
            hue=country_counts.index, palette='Reds_r', legend=False)

plt.title('Dataset Geographic Bias: Top 10 Countries', fontsize=16)
plt.xlabel('Number of Records')
plt.ylabel('Country Code')

# Add percentage labels
total_records = len(df)
for i, count in enumerate(country_counts.values):
    pct = (count / total_records) * 100
    plt.text(count + 2, i, f"{pct:.1f}%", va='center')

plt.show()
```

```python
# Calculate experience representation
exp_counts = df['experience_level'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(exp_counts, labels=exp_counts.index, autopct='%1.1f%%', 
        colors=sns.color_palette('pastel'), startangle=140)

plt.title('Dataset Limitation: Experience Level Imbalance', fontsize=16)
plt.show()

print("Critical Observation: The dataset is dominated by Senior (SE) and Mid-level (MI) roles.")
print("Entry-level (EN) and Executives (EX) are under-represented, meaning our model is less accurate for those extremes.")
```

##  Final Project Conclusion

### Key Findings
1.  **The US Premium:** Geography is the strongest predictor of salary. US-based companies pay on average **$77,000 (114%) more** than the global average ($144k vs $67k).
2.  **Remote Work is Viable:** Statistical testing (T-Test) confirmed **no significant penalty** for working 100% remotely. In fact, fully remote roles showed a higher median salary than hybrid roles in 2022.
3.  **The "Seniority Jump":** The move from Mid-level (MI) to Senior (SE) represents the largest salary increase (~40%) in a data professional's career.
4.  **Company Size Matters:** Mid-sized and Large companies pay significantly more (~$37k premium) than Startups (Small).

### Recommendations for Job Seekers
* **Target US Remote Roles:** The most lucrative strategy is to work remotely for a US-based entity, leveraging the "US Premium" regardless of residence.
* **Skip Hybrid:** Hybrid roles (50% remote) currently pay the lowest on average, likely tied to local cost-of-living constraints.
* **Growth Path:** Focus on reaching "Senior" (SE) status quickly, as this is where the six-figure salary band stabilizes.

### Project Limitations
* **US-Centric Data:** Over 60% of the dataset is US-based, limiting the accuracy of "Global" generalizations.
* **Inflation:** The analysis uses nominal USD values and does not account for the high inflation rates of 2021-2022.
